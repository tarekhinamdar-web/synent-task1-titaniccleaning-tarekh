```python
# Step 1: Import libraries and load dataset
import pandas as pd

# Load dataset directly from URL (or replace with your local file path)
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

print("Data loaded successfully! Rows and Columns:", df.shape)

# Step 2: Handle missing values
# Fill missing 'Age' values with median age
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing 'Embarked' values with the mode (most common value)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Drop 'Cabin' column due to high percentage of missing values
df = df.drop(columns=['Cabin'])

# Step 3: Remove duplicate rows
df = df.drop_duplicates()

# Step 4: Convert data types
# Convert categorical text features to 'category' data type for better efficiency
df['Sex'] = df['Sex'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')
# Ensure target variable is integer
df['Survived'] = df['Survived'].astype('int')

# Step 5: Rename columns to standard lowercase snake_case
df = df.rename(columns={
    'PassengerId': 'passenger_id',
    'Survived': 'survived',
    'Pclass': 'pclass',
    'Name': 'name',
    'Sex': 'sex',
    'Age': 'age',
    'SibSp': 'sibsp',
    'Parch': 'parch',
    'Ticket': 'ticket',
    'Fare': 'fare',
    'Embarked': 'embarked'
})

# Final Check: Verify remaining missing values
print("\nMissing values check:")
print(df.isnull().sum())

# Step 6: Save cleaned dataset to a new CSV file
df.to_csv('cleaned_titanic_data.csv', index=False)
print("\nSuccess! Cleaned dataset saved as 'cleaned_titanic_data.csv'.")
```
